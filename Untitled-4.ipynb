```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

n = 1000  # number of patients

df = pd.DataFrame({
    "age": np.random.randint(30*365, 65*365, n),
    "height": np.random.normal(165, 10, n).astype(int),
    "weight": np.random.normal(70, 15, n),
    "gender": np.random.choice([1, 2], n),
    "ap_hi": np.random.randint(100, 180, n),
    "ap_lo": np.random.randint(60, 110, n),
    "cholesterol": np.random.choice([1, 2, 3], n, p=[0.6, 0.25, 0.15]),
    "gluc": np.random.choice([1, 2, 3], n, p=[0.7, 0.2, 0.1]),
    "smoke": np.random.choice([0, 1], n, p=[0.8, 0.2]),
    "alco": np.random.choice([0, 1], n, p=[0.85, 0.15]),
    "active": np.random.choice([0, 1], n, p=[0.3, 0.7]),
})

# Target variable: cardio risk increases with BP, cholesterol, weight
df["cardio"] = (
    (df["ap_hi"] > 140) |
    (df["cholesterol"] > 1) |
    (df["weight"] > 80)
).astype(int)
```

```python
def draw_cat_plot():
    # Add overweight column
    bmi = df["weight"] / ((df["height"] / 100) ** 2)
    df["overweight"] = (bmi > 25).astype(int)

    # Normalize cholesterol and glucose
    df["cholesterol"] = (df["cholesterol"] > 1).astype(int)
    df["gluc"] = (df["gluc"] > 1).astype(int)

    # Melt data
    df_cat = pd.melt(
        df,
        id_vars=["cardio"],
        value_vars=["cholesterol", "gluc", "smoke", "alco", "active", "overweight"]
    )

    # Group and count
    df_cat = (
        df_cat
        .groupby(["cardio", "variable", "value"])
        .size()
        .reset_index(name="total")
    )

    # Plot
    fig = sns.catplot(
        data=df_cat,
        x="variable",
        y="total",
        hue="value",
        col="cardio",
        kind="bar"
    ).fig

    return fig
```

```python
def draw_heat_map():
    # Clean incorrect data
    df_heat = df[
        (df["ap_lo"] <= df["ap_hi"]) &
        (df["height"] >= df["height"].quantile(0.025)) &
        (df["height"] <= df["height"].quantile(0.975)) &
        (df["weight"] >= df["weight"].quantile(0.025)) &
        (df["weight"] <= df["weight"].quantile(0.975))
    ]

    # Correlation matrix
    corr = df_heat.corr()

    # Mask upper triangle
    mask = np.triu(np.ones_like(corr, dtype=bool))

    # Plot
    fig, ax = plt.subplots(figsize=(12, 10))
    sns.heatmap(
        corr,
        mask=mask,
        annot=True,
        fmt=".1f",
        center=0,
        square=True,
        linewidths=0.5,
        cbar_kws={"shrink": 0.5},
        ax=ax
    )

    return fig
```
